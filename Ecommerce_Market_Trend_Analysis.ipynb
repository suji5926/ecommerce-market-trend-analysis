```python
df_analysis = df.drop(
    columns=[
        "Order_ID_Check",
        "Calculated_Net_Amount",
        "Net_Amount_Check",
        "Column1"
    ]
)

print("Original shape:", df.shape)
print("Analysis shape:", df_analysis.shape)
print(df_analysis.columns.tolist())
```

```python
import pandas as pd
import numpy as np

# Work with the analysis copy created earlier
print("Dataset shape:", df_analysis.shape)
print("Total orders:", df_analysis["Order_id"].nunique())
print("Duplicate rows:", df_analysis.duplicated().sum())
print("Duplicate order IDs:", df_analysis["Order_id"].duplicated().sum())

# Missing values
print("\nMissing values:")
print(df_analysis.isnull().sum())

# Data types
print("\nData types:")
print(df_analysis.dtypes)
```

```python

df_analysis = df_analysis.copy()

text_cols = df_analysis.select_dtypes(include="object").columns

for col in text_cols:
    df_analysis[col] = df_analysis[col].str.strip()


df_analysis["Order_date"] = pd.to_datetime(
    df_analysis["Order_date"], errors="coerce"
)

df_analysis["Customer_signup_date"] = pd.to_datetime(
    df_analysis["Customer_signup_date"], errors="coerce"
)


print("Invalid order dates:", df_analysis["Order_date"].isna().sum())
print("Invalid signup dates:", df_analysis["Customer_signup_date"].isna().sum())

print("\nNumeric summary:")
display(df_analysis.describe().round(2))
```

```python
# Overall business metrics
revenue_summary = pd.DataFrame({
    "Metric": [
        "Total Orders",
        "Total Net Amount",
        "Average Order Value",
        "Total Quantity Sold",
        "Average Discount (%)",
        "Average Delivery Days"
    ],
    "Value": [
        df_analysis["Order_id"].nunique(),
        df_analysis["Net_amount"].sum(),
        df_analysis["Net_amount"].mean(),
        df_analysis["Total_quantity"].sum(),
        df_analysis["Discount_pct"].mean(),
        df_analysis["Delivery_days"].mean()
    ]
})

display(revenue_summary.round(2))
```

```python
monthly_revenue = (
    df_analysis
    .groupby("Order_month")
    .agg(
        Total_orders=("Order_id", "nunique"),
        Total_revenue=("Net_amount", "sum"),
        Average_order_value=("Net_amount", "mean")
    )
    .round(2)
    .sort_index()
)

display(monthly_revenue)

print("Highest revenue month:")
display(monthly_revenue.nlargest(1, "Total_revenue"))

print("Lowest revenue month:")
display(monthly_revenue.nsmallest(1, "Total_revenue"))
```

```python
market_analysis = (
    df_analysis
    .groupby("Order_month")
    .agg(
        Total_revenue=("Net_amount", "sum"),
        Average_search_interest=("Search_interest_index", "mean"),
        Average_market_growth=("Market_growth_pct", "mean"),
        Average_ad_spend=("Industry_ad_spend_cr", "mean"),
        Average_competitor_price=("Avg_competitor_price", "mean")
    )
    .round(2)
    .sort_index()
)

display(market_analysis)
```

```python
average_net_amount = df_analysis["Net_amount"].mean()

high_value_orders = (
    df_analysis[df_analysis["Net_amount"] > average_net_amount]
    [
        ["Order_id", "Customer_id", "Order_date",
         "Main_category", "Net_amount"]
    ]
    .sort_values("Net_amount", ascending=False)
)

display(high_value_orders.head(20))
```

```python
monthly_growth = monthly_revenue.copy()

monthly_growth["Previous_Month_Revenue"] = (
    monthly_growth["Total_revenue"].shift(1)
)

monthly_growth["Growth_Percentage"] = (
    (
        monthly_growth["Total_revenue"]
        - monthly_growth["Previous_Month_Revenue"]
    )
    / monthly_growth["Previous_Month_Revenue"].replace(0, np.nan)
    * 100
).round(2)

display(monthly_growth)
```

```python
category_ranking = category_analysis.copy()

category_ranking["Revenue_Rank"] = (
    category_ranking["Total_revenue"]
    .rank(method="dense", ascending=False)
    .astype(int)
)

display(category_ranking.sort_values("Revenue_Rank"))
```

```python
state_analysis = (
    df_analysis
    .groupby("State")
    .agg(
        Total_orders=("Order_id", "nunique"),
        Total_revenue=("Net_amount", "sum"),
        Average_order_value=("Net_amount", "mean")
    )
    .round(2)
    .sort_values("Total_revenue", ascending=False)
)

display(state_analysis.head(10))
```

```python
delivery_analysis = (
    df_analysis
    .groupby("Order_status")
    .agg(
        Total_orders=("Order_id", "nunique"),
        Average_delivery_days=("Delivery_days", "mean"),
        Median_delivery_days=("Delivery_days", "median")
    )
    .round(2)
)

display(delivery_analysis)
```

```python
coupon_analysis = (
    df_analysis
    .groupby("Coupon_used")
    .agg(
        Total_orders=("Order_id", "nunique"),
        Average_discount=("Discount_pct", "mean"),
        Total_revenue=("Net_amount", "sum"),
        Average_order_value=("Net_amount", "mean")
    )
    .round(2)
)

display(coupon_analysis)
```

```python
df_analysis["Discount_Group"] = pd.cut(
    df_analysis["Discount_pct"],
    bins=[-np.inf, 0, 10, 20, np.inf],
    labels=["No Discount", "Low Discount", "Medium Discount", "High Discount"],
    include_lowest=True
)

discount_analysis = (
    df_analysis
    .groupby("Discount_Group", observed=False)
    .agg(
        Total_orders=("Order_id", "nunique"),
        Total_revenue=("Net_amount", "sum"),
        Average_order_value=("Net_amount", "mean")
    )
    .round(2)
)

display(discount_analysis)
```

```python
status_analysis = (
    df_analysis
    .groupby("Order_status")
    .agg(
        Total_orders=("Order_id", "nunique"),
        Recorded_net_amount=("Net_amount", "sum")
    )
)

status_analysis["Order_percentage"] = (
    status_analysis["Total_orders"]
    / status_analysis["Total_orders"].sum()
    * 100
).round(2)

display(status_analysis.round(2))

cancelled = (df_analysis["Order_status"] == "Cancelled").sum()
returned = (df_analysis["Order_status"] == "Returned").sum()
total_orders = df_analysis["Order_id"].nunique()

print("Cancellation rate (%):", round(cancelled / total_orders * 100, 2))
print("Return rate (%):", round(returned / total_orders * 100, 2))
```

```python
device_analysis = (
    df_analysis
    .groupby("Device")
    .agg(
        Total_orders=("Order_id", "nunique"),
        Total_revenue=("Net_amount", "sum"),
        Average_order_value=("Net_amount", "mean")
    )
    .round(2)
    .sort_values("Total_orders", ascending=False)
)

display(device_analysis)
```

```python
payment_analysis = (
    df_analysis
    .groupby("Payment_method")
    .agg(
        Total_orders=("Order_id", "nunique"),
        Total_revenue=("Net_amount", "sum")
    )
    .round(2)
)

payment_analysis["Order_percentage"] = (
    payment_analysis["Total_orders"]
    / payment_analysis["Total_orders"].sum()
    * 100
).round(2)

display(payment_analysis.sort_values("Total_orders", ascending=False))
```
